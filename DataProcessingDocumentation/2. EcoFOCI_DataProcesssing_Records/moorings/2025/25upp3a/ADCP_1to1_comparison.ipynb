```python
import xarray as xa
```

```python
xdf_scal = xa.load_dataset('25upp3a_wcp_scal.nc')
xdf_scal_orig = xa.load_dataset('25upp3a_scal_0062m.trimmed.nc')
```

```python
for var_name, values in xdf_scal_orig.items():
    try:
        xa.testing.assert_equal(xdf_scal_orig[var_name], xdf_scal[var_name])
        print(f'{var_name}: good')
    except:
        print(f'{var_name}: bad')
```

```python
xdf_velein = xa.load_dataset('25upp3a_wcp_velein.nc')
xdf_velein_orig = xa.load_dataset('25upp3a_0062m.trimmed.nc')
```

```python
for var_name, values in xdf_velein_orig.items():
    try:
        xa.testing.assert_equal(xdf_velein_orig[var_name], xdf_velein[var_name])
        print(f'{var_name}: good')
    except:
        print(f'{var_name}: bad')
```
