# Multi CTD Retrieval

```python
import pandas as pd

import EcoFOCIpy.io.erddap as erddap
import EcoFOCIpy.plots.sbe_ctd_plots as sbe_ctd_plots
```

```python
url = "http://ecofoci-field.pmel.noaa.gov:8080/erddap"

pathname = '/Users/bell/ActiveProjects/fenny_retrieval/MOCNESS_Cruise_list.csv'
```

```python
df = pd.read_csv(pathname)
```

```python
# df = df.iloc[[25]]
df
```

```python
all_ctds = {}

for _i, cruise in df.iterrows():
    ctd_df = None
    try:
        ctd_df = erddap.erddapCruiseretrieve(url=url,cruiseid=cruise['FOCI'].lower(),qclevel='final')
        ctd_df.columns = [x.split(' ')[0] for x in ctd_df.columns]
        all_ctds.update({cruise['FOCI'].lower():ctd_df})
    except:
        try:
            print(f"Dataset-{cruise['FOCI'].lower()} not found")
        except:
            pass
```

```python
for k in all_ctds.keys():
    print(k)
    all_ctds[k].to_csv(f'{k}.csv')
```
