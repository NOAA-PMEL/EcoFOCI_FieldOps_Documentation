# Timeseries make a variable bad and QC flag as 4

```python
import xarray as xa
import numpy as np
import yaml
import os
```

```python
sample_data_dir = '/Users/bell/ecoraid/2025/Moorings/25ckp1a/working/'
```

```python
varlist = ['oxy_percentsat','oxy_concM','oxy_conc_umkg']
for filename in os.listdir(sample_data_dir):
    if 's16' in filename:
        print(filename)
        xdf = xa.load_dataset(sample_data_dir+filename)
        for var in varlist:
            xdf[var+'_QC'] = xdf[var]*0 + 4
            xdf[var] = xdf[var]*0 + np.nan
        xdf.to_netcdf(sample_data_dir+filename,format="NETCDF3_CLASSIC",encoding={'time':{'units':'days since 1900-01-01'}})
```
