# Mooring Netcdf Meta Update

unify units for erddap driven netcdf

```python
import xarray as xa
import yaml
import os
```

```python
sample_data_dir = '/Users/bell/Programs/EcoFOCIpy/'
user_data_dir = '/Users/bell/ecoraid/2025/Moorings/25ckp1a/'
```

```python
datafile = user_data_dir + 'final_data_cf/'
```

```python
inst_meta_file = sample_data_dir+'staticdata/instr_metaconfig/rcmsg_cf.yaml'
```

```python
#instrument info
with open(inst_meta_file) as file:
    inst_config = yaml.full_load(file)
```

```python
inst_config
```

```python
for filename in os.listdir(datafile):
    if 'nc' in filename:
        print(filename)
        if any(sub in filename for sub in ['sg']):
            print(f'*{filename}*')
            xdf = xa.load_dataset(datafile+filename)
            for var in xdf.variables:
                try:
                    xdf[var].attrs = inst_config[var]
                except KeyError:
                    pass
                    print(f'No {var} key in config file')
            xdf.to_netcdf(datafile+filename,format="NETCDF3_CLASSIC",encoding={'time':{'units':'days since 1900-01-01'}})
```
